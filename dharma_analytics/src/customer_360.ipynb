# Default notebook

This default notebook is executed using a Lakeflow job as defined in resources/sample_job.job.yml.

```python
# Set default catalog and schema
catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")
spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{schema}")
spark.sql(f"USE CATALOG {catalog}")
spark.sql(f"USE SCHEMA {schema}")
```

```python
spark.sql(f"""create or replace table {catalog}.{schema}.customer_360 as
          select country,
          count(*)as number_of_customers,
          COUNT_IF (gender = 'female') as  female_customers,
          COUNT_IF(  gender = 'male'  ) as  male_customers
          from samples.bakehouse.sales_customers

          group by country""")
```
